```python
import random, re

text = """The sun always shines brighter after the rain. Every cloud has a silver lining. Look for the good in every day."""

text = re.sub(r'[^ɑ-\u02AF\w\s]', '', text.lower())
words = text.split()

markov = {}

for i in range(len(words)-2):
    key = (words[i], words[i+1])
    nxt = words[i+2]
    markov.setdefault(key, []).append(nxt)

def generate(n=20):
    key = random.choice(list(markov.keys()))
    out = [key[0], key[1]]

    for _ in range(n):
        if key not in markov: # Added check to prevent KeyError
            break
        nxt = random.choice(markov[key])
        out.append(nxt)
        key = (key[1], nxt)

    return " ".join(out)

print(generate(40))
```
